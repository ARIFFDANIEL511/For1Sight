```python

# ============================================================
# FOR1SIGHT V2.8 - STAGE 6
# CELL 1: DEPLOYMENT FILE VERIFICATION
# ============================================================

from pathlib import Path
import json
import joblib
import pandas as pd

# Locate project root
PROJECT_ROOT = Path.cwd()

while (
    PROJECT_ROOT != PROJECT_ROOT.parent
    and not (PROJECT_ROOT / "for1sight_warehouse").exists()
):
    PROJECT_ROOT = PROJECT_ROOT.parent

MODEL_DIR = PROJECT_ROOT / "models" / "v2_8"

WAREHOUSE_PATH = (
    PROJECT_ROOT
    / "for1sight_warehouse"
    / "prediction_training_v2_8_prerace.parquet"
)

print("=" * 65)
print("FOR1SIGHT V2.8 - DEPLOYMENT VERIFICATION")
print("=" * 65)

print("\nProject root:", PROJECT_ROOT)

# 1. Check deployment model files
model_files = {
    "Logistic Regression":
        "logistic_regression_deployment_v2_8.joblib",

    "Random Forest":
        "random_forest_deployment_v2_8.joblib",

    "Gradient Boosting":
        "gradient_boosting_deployment_v2_8.joblib"
}

for name, filename in model_files.items():

    path = MODEL_DIR / filename

    print(f"\n{name}:")
    print("Exists:", path.exists())

    if path.exists():
        print("Size (MB):", round(path.stat().st_size / 1024**2, 2))

# 2. Check metadata
metadata_path = MODEL_DIR / "model_metadata.json"

print("\nMetadata exists:", metadata_path.exists())

if metadata_path.exists():

    with open(metadata_path, "r", encoding="utf-8") as f:
        metadata = json.load(f)

    print("\nTraining seasons:", metadata.get("deployment_training_seasons"))
    print("Training records:", metadata.get("training_records"))
    print("Feature count:", metadata.get("feature_count"))
    print("Features:", metadata.get("features"))

# 3. Inspect model pipelines
print("\n" + "=" * 65)
print("MODEL PIPELINE INSPECTION")
print("=" * 65)

for name, filename in model_files.items():

    path = MODEL_DIR / filename

    if path.exists():

        model = joblib.load(path)

        print("\n", name)
        print("Model class:", type(model).__name__)

        print(
            "Input features:",
            list(model.feature_names_in_)
            if hasattr(model, "feature_names_in_")
            else "Not exposed"
        )

# 4. Inspect warehouse
print("\n" + "=" * 65)
print("WAREHOUSE INSPECTION")
print("=" * 65)

print("Warehouse exists:", WAREHOUSE_PATH.exists())

if WAREHOUSE_PATH.exists():

    warehouse = pd.read_parquet(WAREHOUSE_PATH)

    print("Dataset shape:", warehouse.shape)

    print("\nAvailable columns:")
    print(warehouse.columns.tolist())

    print("\nSeason distribution:")
    print(warehouse["year"].value_counts().sort_index())

print("\n" + "=" * 65)
print("DIAGNOSTIC COMPLETED")
print("=" * 65)
```
