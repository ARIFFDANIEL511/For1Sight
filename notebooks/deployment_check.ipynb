```python
from pathlib import Path
import joblib
import sklearn

MODEL_DIR = Path("models/v2_8")

print("Scikit-learn version:", sklearn.__version__)

models = [
    "gradient_boosting_deployment_v2_8.joblib",
    "logistic_regression_deployment_v2_8.joblib",
    "random_forest_deployment_v2_8.joblib"
]

for filename in models:

    path = MODEL_DIR / filename

    print("\nModel:", filename)
    print("Exists:", path.exists())

    if path.exists():
        try:
            model = joblib.load(path)
            print("Loading: SUCCESS")
            print("Model type:", type(model).__name__)

        except Exception as error:
            print("Loading: FAILED")
            print("Error:", error)
```

```python

from pathlib import Path
import joblib
import sklearn

PROJECT_ROOT = Path(
    r"C:\Users\ayepd\Desktop\For1Sight Project"
)

MODEL_DIR = PROJECT_ROOT / "models" / "v2_8"

MODEL_FILES = {
    "Gradient Boosting":
        "gradient_boosting_deployment_v2_8.joblib",

    "Logistic Regression":
        "logistic_regression_deployment_v2_8.joblib",

    "Random Forest":
        "random_forest_deployment_v2_8.joblib"
}

print("=" * 55)
print("FOR1SIGHT V2.8 - MODEL VERIFICATION")
print("=" * 55)

print("Scikit-learn:", sklearn.__version__)
print("Model directory:", MODEL_DIR)

loaded_models = {}

for model_name, filename in MODEL_FILES.items():

    model_path = MODEL_DIR / filename

    print(f"\nChecking {model_name}...")

    if not model_path.exists():
        print("FILE NOT FOUND")
        continue

    try:
        model = joblib.load(model_path)

        loaded_models[model_name] = model

        print("File exists: YES")
        print("Model loaded: SUCCESS")
        print("Model type:", type(model).__name__)

        if hasattr(model, "feature_names_in_"):
            print(
                "Features:",
                list(model.feature_names_in_)
            )

    except Exception as error:
        print("MODEL LOADING FAILED:", error)

print("\n" + "=" * 55)
print("SUCCESSFULLY LOADED:",
      len(loaded_models), "/ 3 MODELS")
print("=" * 55)
```
