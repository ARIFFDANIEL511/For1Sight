```python
import pandas as pd

df = pd.read_parquet(
    parquet_path,
    engine="pyarrow"
)

print("Silver Parquet successfully loaded!")

print("Dataset shape:", df.shape)

print("\nAvailable columns:")
print(df.columns.tolist())
```

```python

# ==========================================================
# FOR1SIGHT — SILVER PARQUET DATA DIAGNOSTICS
# INVESTIGATING CIRCUIT PERFORMANCE FILTERING
# ==========================================================

import pandas as pd

print("=" * 65)
print("FOR1SIGHT SILVER DATA DIAGNOSTICS")
print("=" * 65)

# 1. Check data types
print("\n1. COLUMN DATA TYPES")

columns_to_check = [
    "year", "event_name", "session_type",
    "driver_code", "lap", "time",
    "s1", "s2", "s3", "del", "iacc"
]

for col in columns_to_check:
    if col in df.columns:
        print(f"{col:15} : {df[col].dtype}")

# 2. Check season distribution
print("\n2. RECORDS BY YEAR")
print(df["year"].value_counts(dropna=False))

# 3. Check event names by season
print("\n3. 2024 EVENT NAMES")

year_2024 = df[
    df["year"].astype(str).str.strip() == "2024"
]

print("2024 record count:", len(year_2024))
print("\nAvailable events:")
print(sorted(year_2024["event_name"].dropna().unique()))

# 4. Check session types
print("\n4. SESSION TYPES")
print(year_2024["session_type"].value_counts(dropna=False))

# 5. Check driver codes
print("\n5. DRIVER CODES — 2024")
print(year_2024["driver_code"].value_counts().head(30))

# 6. Inspect lap-time formats
print("\n6. LAP TIME SAMPLE")

for col in ["time", "s1", "s2", "s3"]:
    print(f"\n{col}:")
    print(year_2024[col].dropna().head(8).map(repr).to_string())

# 7. Investigate quality flags
print("\n7. QUALITY FLAG DISTRIBUTION")

for col in ["del", "iacc"]:
    print(f"\n{col}:")
    print(year_2024[col].value_counts(dropna=False))

# 8. Check missing values
print("\n8. MISSING VALUES")

for col in columns_to_check:
    if col in df.columns:
        print(
            f"{col:15} : {df[col].isna().sum():,}"
        )

print("\n" + "=" * 65)
print("DIAGNOSTICS COMPLETED")
print("=" * 65)
```

```python

# ==========================================================
# FOR1SIGHT — DRIVER COMPARISON FILTER TEST
# ==========================================================

test_df = df.copy()

# Selected dashboard settings
selected_year = 2024
selected_event = "Singapore Grand Prix"
selected_session = "Race"

selected_drivers = ["ALB", "ALO"]

# 1. Filter season
test_df = test_df[test_df["year"] == selected_year]

print("Records after season filter:", len(test_df))

# 2. Filter circuit
test_df = test_df[
    test_df["event_name"] == selected_event
]

print("Records after circuit filter:", len(test_df))

# 3. Filter session
test_df = test_df[
    test_df["session_type"] == selected_session
]

print("Records after session filter:", len(test_df))

# 4. Filter selected drivers
test_df = test_df[
    test_df["driver_code"].isin(selected_drivers)
].copy()

print("Records after driver filter:", len(test_df))

# 5. Apply correct quality filters
deleted = (
    test_df["del"].astype(str).str.lower().str.strip()
    .isin(["true", "1", "yes"])
)

inaccurate = (
    test_df["iacc"].astype(str).str.lower().str.strip()
    .isin(["false", "0", "no"])
)

test_df = test_df[~deleted & ~inaccurate].copy()

print("Records after quality filtering:", len(test_df))

# 6. Display results by driver
print("\nFINAL RECORDS BY DRIVER")

print(
    test_df.groupby("driver_code")
    .agg(
        total_laps=("lap", "count"),
        average_lap_time=("time", "mean"),
        best_lap_time=("time", "min"),
        average_sector_1=("s1", "mean"),
        average_sector_2=("s2", "mean"),
        average_sector_3=("s3", "mean")
    )
)

print("\nFILTER TEST COMPLETED")
```
